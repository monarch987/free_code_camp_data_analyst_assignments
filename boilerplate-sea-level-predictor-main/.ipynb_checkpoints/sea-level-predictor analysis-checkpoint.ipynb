```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import scipy as sp
```

```python
%matplotlib inline
```

```python
df = pd.read_csv('epa-sea-level.csv')
df.head(-5)
```

## SCATTER PLOT

```python
#using matplotlib
x = np.array(df['Year'])
y = np.array(df['CSIRO Adjusted Sea Level'])
# plt.scatter(x=df['Year'], y=df['CSIRO Adjusted Sea Level'])
plt.scatter(x, y)
plt.xlabel("Year")
plt.ylabel("CSIRO Adjusted Sea Level")
plt.show()
```

```python
sns.scatterplot(x='Year',y='CSIRO Adjusted Sea Level',data=df)
```

## BEST-FIT

```python
#finding slope and intercept of the best fit line using linregress 
slope,intercept, r_value, p_value, std_err = sp.stats.linregress(x=df['Year'],y=df['CSIRO Adjusted Sea Level'])
slope,intercept
```

```python
#plotting using matplotlib

x_pred = np.arange(df['Year'].min(),2051,1)
y_pred = slope*x_pred+intercept

plt.scatter(x_pred,y_pred)
plt.plot(x_pred,y_pred,color='red',label=f'y={slope}x+{intercept}')

plt.xlabel("Year")
plt.ylabel("CSIRO Adjusted Sea Level")
plt.show()
```

```python
#plotting using seaborn

x_pred = np.arange(df['Year'].min(),2051)
y_pred = slope*x_pred + intercept 

sns.regplot(
    x_pred,
    y_pred,
    line_kws={"color": "red", "label": f"Fit: y = {slope:.2f}x + {intercept:.2f}"}
)
```

```python
#2000 year onward (calculate the slope & intercept and plot upto year 2050 )

predicted_df = df[df['Year']>2000]

val = sp.stats.linregress(x=predicted_df['Year'],y=predicted_df['CSIRO Adjusted Sea Level'])
val.slope,val.intercept
```

```python
x_pred = np.arange(predicted_df['Year'].min(),2051)
y_pred = val.slope*x_pred + val.intercept 

sns.regplot(
    x_pred,
    y_pred,
    line_kws={"color": "red", "label": f"Fit: y = {slope:.2f}x + {intercept:.2f}"}
)
```
