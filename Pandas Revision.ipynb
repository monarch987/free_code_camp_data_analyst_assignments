## conditionals (>,<,|,&)

```python
pdf = pd.DataFrame(randn(5,3),['a','b','c','d','e'],['f','g','h'])
pdf
```

```python
val_1 = pdf['g']>0
```

```python
pdf[val_1]
```

```python
val_2 = (pdf['g']>0) & (pdf['h']<0)  

pdf[val_2]
```

```python
val_3 = (pdf['g']>0) | (pdf['h']>0)  

pdf[val_3]
```

## Handling Missing Data

```python
dic = {'A':[1,2,np.nan],'B':[3,np.nan,np.nan],'C':[1,3,4]}

DF = pd.DataFrame(dic)
DF
```

```python
## droping column having null values

DF_dropcol = DF.dropna(axis=1)
DF_dropcol
```

```python
## droping rows having null values

DF_droprow = DF.dropna(axis=0)
DF_droprow
```

```python
# droping rows having more than 1 null values (same applies to column by making axis=1)

DF_droprow_thresh = DF.dropna(axis=0,thresh=2)
DF_droprow_thresh
```

```python
## filling up null values with a default value

DF_default = DF.fillna(value=0)
DF_default
```

```python
## filling up formulated values in place of null values

DF_custom_value = DF.fillna(DF['A'].mean()) 
DF_custom_value
```

## GROUPBY

```python
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
        'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
        'Sales':[200,120,340,124,243,350]}

Df = pd.DataFrame(data)
Df
```

```python
#find out common  value on multiple rows for a given column(Company in our case) and group them 
company = Df.groupby('Company')
company
```

```python
##mean of avail col

company.mean('Sales')
```

```python
#summing up values in a group for column ('Sales')
company.sum('Sales')
```

```python
#to locate a specific cell 

company.sum('Sales').loc['GOOG']
```

```python
#min sales
company.min('Sales')
```

```python
company.max('Sales')
```
