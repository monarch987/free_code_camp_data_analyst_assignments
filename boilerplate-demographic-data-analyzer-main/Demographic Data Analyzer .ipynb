```python
data = pd.read_csv('adult_data.csv')

data.head(-1)
```

```python
#How many people of each race are represented in this dataset? 

race = data.groupby('race').size()
race
# print(type(race))
```

```python
#What is the average age of men?
avg_age = data.groupby('sex')['age'].mean()
avg_age.loc['Male']
# print(type(avg_age))
```

```python
#What is the percentage of people who have a Bachelor's degree?

#pandas-series
education = data.groupby('education').size()

#summation of column education
total_populus = education.sum()

percent_bachelor = (education['Bachelors']/total_populus)*100
percent_bachelor
# print(total_populus)
```

```python
#What percentage of people with advanced education (Bachelors, Masters, or Doctorate) make more than 50K?

#data-frame
greater_than_50k = data[data['salary']=='>50K']

#series
seria_a = greater_than_50k.groupby('education').size()

total_educated_populus = data.groupby('education').size().sum()


percent_adv_bachelor = (seria_a['Bachelors']/total_educated_populus)*100

percent_adv_master = (seria_a['Masters']/total_educated_populus)*100

percent_adv_Doctorate = (seria_a['Doctorate']/total_educated_populus)*100

# percent_adv_Doctorate
```

```python
#What percentage of people without advanced education make more than 50K?

non_advanced_greater_than_50k = data[data['salary']=='>50K']
non_adv_occupation  = non_advanced_greater_than_50k[~non_advanced_greater_than_50k['education'].isin(['Bachelors','Masters','Doctorate'])]
non_adv_occupation.groupby('education').size()
```

```python
#What percentage of people without advanced education make more than 50K?

total_adv_edu_percent_greater_than_50k = percent_adv_bachelor+percent_adv_master+percent_adv_Doctorate

non_adv_percent_greater_than_50k = 100 - total_adv_edu_percent_greater_than_50k

non_adv_percent_greater_than_50k
```

```python
#What is the minimum number of hours a person works per week?

minimum_no_hrs = data['hours-per-week'].min()

minimum_no_hrs
```

```python
#What percentage of the people who work the minimum number of hours per week have a salary of more than 50K?

no_of_people_slaried_greater_than_50k = data[data['salary']=='>50K']

working_min_hrs = no_of_people_slaried_greater_than_50k[no_of_people_slaried_greater_than_50k['hours-per-week']==1]
len(working_min_hrs)
```

```python
#What country has the highest percentage of people that earn >50K and what is that percentage?

#pandas-series
country_tops = no_of_people_slaried_greater_than_50k.groupby('native-country').size()

# print(country_tops)
summation_of_all_country = country_tops.sum()

# print(summation_country_tops)

max_people = country_tops.max()

#whats the percent shar
percent = (max_people/summation_of_all_country)*100


#which country tops the list
country = country_tops[country_tops == max_people].idxmax()
# country
print(percent,country)
```

```python
#Identify the most popular occupation for those who earn >50K in India.

#dataframe
indian_populus = no_of_people_slaried_greater_than_50k[no_of_people_slaried_greater_than_50k['native-country']=='India']

#series
occupation_wise = indian_populus.groupby('occupation').size()
# print(occupation_wise)

most_prefered_occupation_count = occupation_wise.max()

most_prefered_occupation = occupation_wise[occupation_wise==most_prefered_occupation_count].idxmax()
most_prefered_occupation
```
